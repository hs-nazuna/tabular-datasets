```python
import numpy as np
import pandas as pd
from ucimlrepo import fetch_ucirepo

data = fetch_ucirepo(id=560)
X = data.data.features.drop(columns=['Date'])
y = data.data.targets
print(X.shape)
```

```python
for i, col in enumerate(X.columns):
    if X[col].dtype == 'object':
        X.iloc[:,i] = X[col].replace('?', np.nan)

df = pd.concat([X, y], axis=1)
df = df.dropna().reset_index(drop=True)
y = df['Rented Bike Count']
df = df.drop(columns=['Rented Bike Count'])
df['Rented Bike Count'] = y
df['Functioning Day'] = df['Functioning Day'].map({'Yes': 1, 'No': 0})
df.to_csv('seoul-bike-sharing.csv', index=False)
df.shape
```
